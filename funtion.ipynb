```python
def fact (n):
  f = 1
  for x in range(1,n+1,1):
      f= f*x
  return f
print(fact(5),fact(7),fact(10))
```

```python
def fact (n):
  f = 1
  x = 1
  while x in range(x<=n):
      f= f*x
      x= x+1
  return f
print(fact(5))
```
